```python
import os

def configure_firewall_rules():

    print("Configurazione regole firewall: filtraggio porta 80")

    os.system("iptables -F")
    os.system("iptables -X")

    os.system("iptables -P FORWARD DROP")

    os.system("iptables -A FORWARD -m state --state ESTABLISHED,RELATED -j ACCEPT")

    os.system("iptables -A FORWARD -p tcp --dport 80 -d 10.0.3.10 -j ACCEPT")

    os.system("iptables -A FORWARD -p icmp -j ACCEPT")

    print("[OK] Regole iptables applicate correttamente.")

if __name__ == "__main__":
    
    configure_firewall_rules()
```
